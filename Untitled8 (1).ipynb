```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.metrics import r2_score

plt.rcParams.update({
    "text.usetex": False,
    "font.family": "serif",
    "font.serif": ["Times New Roman", "DejaVu Serif"],

    "axes.labelsize": 15,
    "axes.titlesize": 17,
    "axes.labelweight": "bold",
    "axes.titleweight": "bold",

    "xtick.labelsize": 13,
    "ytick.labelsize": 13,

    "legend.fontsize": 12,

    "grid.linestyle": "--",
    "grid.alpha": 0.3,

    "figure.dpi": 600
})

# Load saved predictions
faett_df = pd.read_csv(
    "/content/drive/MyDrive/Parkinson_Transformer_Project/results/faett_predictions.csv"
)

y_true = faett_df["Actual"]
y_pred = faett_df["Predicted"]

r2 = r2_score(y_true, y_pred)

plt.figure(figsize=(8,8))

plt.scatter(
    y_true,
    y_pred,
    color="#2E8B57",          # bubble color
    edgecolors="black",       # border color
    linewidths=0.7,
    alpha=0.75,
    s=70
)

# Perfect prediction line
min_val = min(y_true.min(), y_pred.min())
max_val = max(y_true.max(), y_pred.max())

plt.plot(
    [min_val, max_val],
    [min_val, max_val],
    color="#8B0000",          # dark red line
    linestyle="--",
    linewidth=2.5,
    label="Perfect Prediction"
)

plt.text(
    0.05,
    0.95,
    f"$R^2$ = {r2:.4f}",
    transform=plt.gca().transAxes,
    fontsize=13,
    verticalalignment='top',
    bbox=dict(
        facecolor='white',
        edgecolor='black',
        boxstyle='round,pad=0.4'
    )
)

plt.xlabel("Actual Total UPDRS")
plt.ylabel("Predicted Total UPDRS")

#plt.title(
 #   "FAETT: Actual vs Predicted Values"
#)

plt.legend(frameon=True)

plt.grid(True)

plt.tight_layout()

plt.savefig(
    "/content/drive/MyDrive/Parkinson_Transformer_Project/results/FAETT_Actual_vs_Predicted.png",
    dpi=600,
    bbox_inches="tight"
)

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load saved history
history_df = pd.read_csv(
    "/content/drive/MyDrive/Parkinson_Transformer_Project/results/transformer_history.csv"
)

plt.rcParams.update({
    "text.usetex": False,
    "font.family": "serif",
    "font.serif": ["Times New Roman", "DejaVu Serif"],

    "axes.labelsize": 15,
    "axes.titlesize": 17,
    "axes.labelweight": "bold",
    "axes.titleweight": "bold",

    "xtick.labelsize": 13,
    "ytick.labelsize": 13,

    "legend.fontsize": 12,

    "grid.linestyle": "--",
    "grid.alpha": 0.3,

    "figure.dpi": 600
})

plt.figure(figsize=(10,6))

plt.plot(
    history_df["loss"],
    color="#8B0000",
    linewidth=2.5,
    label="Training Loss"
)

plt.plot(
    history_df["val_loss"],
    color="#2E8B57",
    linewidth=2.5,
    label="Validation Loss"
)

plt.xlabel("Epoch")
plt.ylabel("Loss (MSE)")

plt.legend(
    frameon=True,
    edgecolor="black"
)

plt.grid(True)

plt.tight_layout()

plt.savefig(
    "/content/drive/MyDrive/Parkinson_Transformer_Project/results/Transformer_Training_Validation_Loss.png",
    dpi=600,
    bbox_inches="tight"
)

plt.show()
```
